## 1. Fundamentos Teóricos

El **problema de los dos cuerpos** (Sol + Apophis) admite exactamente **seis constantes de movimiento**:
tres componentes del momento angular $\\mathbf{h}$ y tres del vector LRL $\\mathbf{e}$, más la energía
específica $\\varepsilon$ (que no es independiente). En total, para la solución de la órbita elíptica,
las tres integrales primeras independientes son:

$$\\boxed{\\mathbf{h}, \\quad \\varepsilon, \\quad \\mathbf{e}_{\\text{LRL}}}$$

Estas integrales primeras **solamente existen en el potencial $-\\mu/r$** (gravitación pura). Cualquier
perturbación — paso cercano a la Tierra, presión de radiación solar, mareas, o el propio impacto DART —
las quiebra de forma medible.

### 1.1 MARE — Momento Angular Relativo Específico

El **momento angular relativo específico** se define como el producto vectorial del radio con la velocidad:

$$\\mathbf{h} = \\mathbf{r} \\times \\mathbf{v}$$

Su magnitud satisface la relación kepleriana:

$$|\\mathbf{h}| = \\sqrt{\\mu p}, \\qquad p = a(1-e^2)$$

donde $p$ es el **semilatus rectum**, $a$ el semieje mayor y $e$ la excentricidad.

**¿Por qué se conserva?**  
La ecuación de movimiento kepleriana es:

$$\\ddot{\\mathbf{r}} = -\\frac{\\mu}{r^3}\\mathbf{r}$$

Tomando la derivada temporal de $\\mathbf{h}$:

$$\\frac{d\\mathbf{h}}{dt} = \\dot{\\mathbf{r}} \\times \\mathbf{v} + \\mathbf{r} \\times \\ddot{\\mathbf{r}}
= \\mathbf{v} \\times \\mathbf{v} + \\mathbf{r} \\times \\left(-\\frac{\\mu}{r^3}\\mathbf{r}\\right)
= \\mathbf{0} + \\mathbf{0} = \\mathbf{0}$$

La primera componente es cero por antisimetría del producto vectorial; la segunda porque
$\\mathbf{r} \\times \\mathbf{r} = \\mathbf{0}$. La fuerza central **no tiene torque**.

**Significado físico:**
- $\\mathbf{h}$ es **perpendicular al plano orbital** (lo define).
- Su magnitud es el **doble de la velocidad areolar**: $|\\mathbf{h}| = 2\\,\\dot{A}$ (segunda ley de Kepler).
- Para Apophis: $|\\mathbf{h}| \\approx \\sqrt{\\mu a(1-e^2)} \\approx \\sqrt{1 \\cdot 0.922 \\cdot (1-0.1914^2)} \\approx 0.953$ AU²/UT.

### 1.2 ERE — Energía Relativa Específica

La **energía mecánica específica** (cinética + potencial por unidad de masa reducida) es:

$$\\varepsilon = \\frac{v^2}{2} - \\frac{\\mu}{r}$$

Para una órbita kepleriana, usando la ecuación de la vis-viva, $\\varepsilon$ se relaciona con $a$:

$$\\boxed{\\varepsilon = -\\frac{\\mu}{2a}}$$

| Tipo de órbita | Signo de $\\varepsilon$ | Semieje $a$ |
|----------------|------------------------|-------------|
| Elipse         | $\\varepsilon < 0$     | $a > 0$     |
| Parábola       | $\\varepsilon = 0$     | $a \\to \\infty$ |
| Hipérbola      | $\\varepsilon > 0$     | $a < 0$     |

**¿Por qué se conserva?**  
El potencial gravitacional $U = -\\mu/r$ es **conservativo** (función solo de $r$, independiente de
$\\mathbf{v}$ y $t$). Por tanto:

$$\\frac{d\\varepsilon}{dt} = \\mathbf{v} \\cdot \\ddot{\\mathbf{r}} - \\frac{\\mu}{r^2}\\dot{r}
= \\mathbf{v} \\cdot \\left(-\\frac{\\mu}{r^3}\\mathbf{r}\\right) + \\frac{\\mu}{r^2} \\frac{\\mathbf{r}\\cdot\\mathbf{v}}{r} = 0$$

**Importancia para DART:** Al momento del impacto $t_\\text{dart}$, el objeto recibe un impulso $\\Delta\\mathbf{V}$
y la energía cambia **instantáneamente** en:

$$\\Delta\\varepsilon = \\mathbf{v} \\cdot \\Delta\\mathbf{V} + \\frac{|\\Delta\\mathbf{V}|^2}{2}$$

Este salto determina el nuevo semieje mayor $a' = -\\mu/(2\\varepsilon')$ y, por ende, el nuevo período orbital.

### 1.3 Vector de Laplace-Runge-Lenz (LRL)

El **vector de Laplace-Runge-Lenz** es la constante de movimiento más profunda del problema kepleriano:

$$\\mathbf{e} = \\frac{\\mathbf{v} \\times \\mathbf{h}}{\\mu} - \\hat{\\mathbf{r}}$$

donde $\\hat{\\mathbf{r}} = \\mathbf{r}/|\\mathbf{r}|$ es el vector unitario radial.

**Propiedades fundamentales:**
1. $|\\mathbf{e}| = e$ (módulo igual a la excentricidad).
2. $\\mathbf{e}$ **apunta al periapsis** (dirección fija en el espacio inercial).
3. $\\mathbf{e} \\perp \\mathbf{h}$ (vive en el plano orbital).
4. La ecuación de la cónica se recupera directamente: $r = p/(1 + e\\cos\\theta)$.

**¿Por qué se conserva?** En el potencial $\\propto 1/r$, la simetría adicional (simetría SO(4) para
$\\varepsilon < 0$) garantiza la existencia de este invariante extra:

$$\\frac{d\\mathbf{e}}{dt} = \\mathbf{0} \\quad \\text{(para potencial puramente kepleriano)}$$

**Cuando se rompe la conservación:**
- **Impacto DART:** $\\mathbf{e}$ cambia de módulo ($\\Delta e$) y de dirección ($\\Delta\\theta_{\\text{peri}}$).
  El periapsis se desplaza, acercando o alejando el punto de máxima velocidad de la órbita terrestre.
- **Perturbación terrestre:** Durante el encuentro 2029, la tracción gravitacional de la Tierra modifica
  $\\mathbf{e}$ de forma continua, lo que se observa como una leve precesión del periapsis.
- **Presión de radiación / efecto Yarkovsky:** Perturbaciones no gravitacionales que actúan sobre escalas
  de años y producen una deriva secular medible en $\\mathbf{e}$.

---

### 1.4 Importancia para la misión DART inverso

| Invariante | Antes del impacto | Después del impacto | Observable |
|------------|-------------------|---------------------|------------|
| $|\\mathbf{h}|$ | constante | cambia en $|\\mathbf{r} \\times \\Delta\\mathbf{V}|$ | inclinación, $\\Omega$, $\\omega$ |
| $\\varepsilon$ | constante | salto $\\Delta\\varepsilon = \\mathbf{v}\\cdot\\Delta\\mathbf{V} + |\\Delta\\mathbf{V}|^2/2$ | semieje mayor $a'$, período $T'$ |
| $\\mathbf{e}_{\\text{LRL}}$ | constante | cambia módulo y dirección | excentricidad $e'$, argumento del periapsis $\\omega'$ |

La **estrategia inversa** al DART consiste en aplicar $\\Delta\\mathbf{V}$ en la dirección y magnitud exactas
que garanticen que Apophis **no impacte la Tierra en abril 2029**, minimizando el costo energético
$|\\Delta\\mathbf{V}|$. Este cuaderno valida el punto de partida: las constantes de movimiento kepleriano
son efectivamente constantes antes del impacto.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

import pymcel as pc
from pymcel.constantes import GM_sun, au, year

# ── Canonical units (Sun-centric, AU-based) ─────────────────────────────────
# UL = 1 AU = 1.496e11 m
# UM = M_sun
# UT derived from Kepler's 3rd law: T_earth = 2π → UT = T_earth/(2π)
UL_m  = au          # 1 AU in metres
UT_s  = np.sqrt(UL_m**3 / GM_sun)  # 1 UT in seconds (≈ 5.022e6 s)
UT_days = UT_s / 86400             # 1 UT in days   (≈ 58.13 days)

mu_sol = 1.0   # [AU³/UT²] in canonical units

print(f"Unidades canónicas:")
print(f"  UL = 1 AU = {UL_m:.4e} m")
print(f"  UT = {UT_s:.4e} s  =  {UT_days:.4f} días")
print(f"  UV = UL/UT = {UL_m/UT_s:.4f} m/s  (velocidad canónica)")
print(f"  μ_Sol = {mu_sol} [AU³/UT²]")
```

## 2. Obtención de Efemérides de Apophis

Consultamos el sistema Horizons de la NASA para obtener el estado cartesiano $(\\mathbf{r}, \\mathbf{v})$
de Apophis (ID `99942`) referenciado al centro solar (`@0`), en el sistema eclíptico J2000,
para el intervalo **2028-01-01 → 2029-06-01** con paso diario.

Las velocidades vienen en km/s desde Horizons; las convertimos a unidades canónicas:

$$v_{\\text{can}} = v_{\\text{km/s}} \\times \\frac{\\text{UT}_{\\text{días}} \\times 86400}{\\text{UL}_{\\text{km}}}$$

```python
epochs = {'start': '2028-01-01', 'stop': '2029-06-01', 'step': '1d'}

df = pc.consulta_horizons(
    id='99942',
    location='@0',
    epochs=epochs,
    datos='vectors'
)

print(f"Filas obtenidas: {len(df)}")
print(f"Columnas: {list(df.columns)}")
df.head(3)
```

```python
# Horizons returns positions in AU and velocities in AU/day
# Convert velocities: AU/day → AU/UT
UV_AUperDay_to_can = UT_days  # multiply km/s by UT_days*86400/UL_km, but
                               # Horizons vectors give v in AU/day; 1 AU/day = UT_days AU/UT

# Detect velocity column units from Horizons output
# pc.consulta_horizons with datos='vectors' returns vx,vy,vz in AU/day
df['vx_can'] = df['vx'] * UT_days
df['vy_can'] = df['vy'] * UT_days
df['vz_can'] = df['vz'] * UT_days

# Parse dates
dates = pd.to_datetime(df['datetime_str'])

print("Muestra de estado canónico:")
print(df[['datetime_str','x','y','z','vx_can','vy_can','vz_can']].head(3).to_string(index=False))
```

## 3. Cálculo de los Invariantes a lo Largo de la Órbita

Para cada punto $(\\mathbf{r}_i, \\mathbf{v}_i)$ de la efeméride calculamos:

$$\\mathbf{h}_i = \\mathbf{r}_i \\times \\mathbf{v}_i, \\qquad
\\varepsilon_i = \\frac{v_i^2}{2} - \\frac{\\mu}{r_i}, \\qquad
\\mathbf{e}_i = \\frac{\\mathbf{v}_i \\times \\mathbf{h}_i}{\\mu} - \\hat{\\mathbf{r}}_i$$

Los errores relativos respecto al valor inicial miden cuánto se desvía la integración numérica
de la conservación exacta.

```python
r_arr = df[['x', 'y', 'z']].values                        # (N,3) AU
v_arr = df[['vx_can', 'vy_can', 'vz_can']].values         # (N,3) AU/UT

# ── MARE: h = r × v ────────────────────────────────────────────────────────
h_arr = np.cross(r_arr, v_arr)                             # (N,3)
h_mag = np.linalg.norm(h_arr, axis=1)                     # (N,)

# ── ERE: ε = v²/2 - μ/r ────────────────────────────────────────────────────
r_mag = np.linalg.norm(r_arr, axis=1)
v_mag = np.linalg.norm(v_arr, axis=1)
eps_arr = v_mag**2 / 2 - mu_sol / r_mag

# ── LRL: e_vec = (v × h)/μ - r_hat ─────────────────────────────────────────
r_hat   = r_arr / r_mag[:, None]
e_vec   = np.cross(v_arr, h_arr) / mu_sol - r_hat         # (N,3)
e_mag   = np.linalg.norm(e_vec, axis=1)                   # (N,)

# ── Reference values at t=0 ─────────────────────────────────────────────────
h0   = h_mag[0]
eps0 = eps_arr[0]
e0   = e_mag[0]

# ── Relative errors ─────────────────────────────────────────────────────────
dh   = (h_mag   - h0) / h0
deps = (eps_arr - eps0) / abs(eps0)
de   = (e_mag   - e0) / e0

# ── Orbital elements cross-check via pymcel ──────────────────────────────────
estado0 = np.concatenate([r_arr[0], v_arr[0]])
elems0  = pc.estado_a_elementos(mu_sol, estado0)
p0, e_elem, inc, Omega, omega, f = elems0
a0 = p0 / (1 - e_elem**2)

print("Verificación de conservación a lo largo de la órbita de Apophis:")
print(f"  h₀      = {h0:.10f}  AU²/UT")
print(f"  ε₀      = {eps0:.10f}  AU²/UT²")
print(f"  e₀      = {e0:.10f}  (sin unidades)")
print(f"  a₀      = {a0:.6f}  AU   (de vis-viva: {-mu_sol/(2*eps0):.6f})")
print(f"  e (pymcel) = {e_elem:.10f}")
print()
print(f"  |Δh/h₀|_max   = {np.abs(dh).max():.2e}")
print(f"  |Δε/ε₀|_max   = {np.abs(deps).max():.2e}")
print(f"  |Δe/e₀|_max   = {np.abs(de).max():.2e}")
```

## 7. Salto en Invariantes al Aplicar el Impulso DART

Al aplicar un impulso $\\Delta\\mathbf{V}$ en el instante $t_\\text{dart}$, las tres constantes de
movimiento cambian **de forma instantánea**. Las fórmulas analíticas son:

$$\\Delta\\varepsilon = \\mathbf{v} \\cdot \\Delta\\mathbf{V} + \\frac{|\\Delta\\mathbf{V}|^2}{2}$$

$$\\Delta\\mathbf{h} = \\mathbf{r} \\times \\Delta\\mathbf{V}$$

$$\\mathbf{e}' = \\frac{(\\mathbf{v}+\\Delta\\mathbf{V}) \\times \\mathbf{h}'}{\\mu} - \\hat{\\mathbf{r}}$$

Usando la **ecuación de la vis-viva**, el nuevo semieje mayor queda:

$$a' = -\\frac{\\mu}{2\\varepsilon'} = -\\frac{\\mu}{2(\\varepsilon + \\Delta\\varepsilon)}$$

El código siguiente evalúa estas expresiones para un rango de $|\\Delta\\mathbf{V}|$ aplicado
en dirección **radial** (dirección $\\hat{\\mathbf{r}}$) desde el estado de Apophis 180 días
antes del encuentro — caso representativo de la misión DART inverso.

```python
# State 180 days before end of ephemeris  (proxy for τ = 180 d case)
idx_dart = -181
row = df.iloc[idx_dart]

r_dart = np.array([row['x'],      row['y'],      row['z']])
v_dart = np.array([row['vx_can'], row['vy_can'], row['vz_can']])
r_dart_mag = np.linalg.norm(r_dart)
v_dart_mag = np.linalg.norm(v_dart)
r_dart_hat = r_dart / r_dart_mag
v_dart_hat = v_dart / v_dart_mag

# Before impact
eps_before = v_dart_mag**2 / 2 - mu_sol / r_dart_mag
h_before   = np.cross(r_dart, v_dart)
e_vec_before = np.cross(v_dart, h_before) / mu_sol - r_dart_hat
a_before   = -mu_sol / (2 * eps_before)

# Range of ΔV (canonical units AU/UT)
# 1 AU/UT ≈ UL_m/UT_s m/s; convert to m/s for labelling
UV_ms = UL_m / UT_s  # 1 AU/UT in m/s
dv_max_ms  = 1000.0  # 1000 m/s sweep for general view
dv_range   = np.linspace(0, dv_max_ms / UV_ms, 400)   # AU/UT
dv_range_ms = dv_range * UV_ms                          # m/s

# Apply ΔV in the radial direction (along r_hat)
eps_after_rad  = np.zeros(len(dv_range))
a_after_rad    = np.zeros(len(dv_range))
dh_mag_rad     = np.zeros(len(dv_range))
de_lrl_rad     = np.zeros(len(dv_range))

# Apply ΔV in the tangential direction (along v_hat)
eps_after_tan  = np.zeros(len(dv_range))
a_after_tan    = np.zeros(len(dv_range))

for i, dv in enumerate(dv_range):
    # Radial impulse
    dV_rad   = dv * r_dart_hat
    v_new_r  = v_dart + dV_rad
    eps_r    = np.dot(v_new_r, v_new_r) / 2 - mu_sol / r_dart_mag
    h_new_r  = np.cross(r_dart, v_new_r)
    e_new_r  = np.cross(v_new_r, h_new_r) / mu_sol - r_dart_hat
    eps_after_rad[i] = eps_r
    a_after_rad[i]   = -mu_sol / (2 * eps_r) if eps_r < 0 else np.nan
    dh_mag_rad[i]    = (np.linalg.norm(h_new_r) - np.linalg.norm(h_before)) / np.linalg.norm(h_before)
    de_lrl_rad[i]    = (np.linalg.norm(e_new_r) - np.linalg.norm(e_vec_before)) / np.linalg.norm(e_vec_before)

    # Tangential impulse
    dV_tan   = dv * v_dart_hat
    v_new_t  = v_dart + dV_tan
    eps_t    = np.dot(v_new_t, v_new_t) / 2 - mu_sol / r_dart_mag
    eps_after_tan[i] = eps_t
    a_after_tan[i]   = -mu_sol / (2 * eps_t) if eps_t < 0 else np.nan

print(f"Estado DART (τ=180 d):")
print(f"  fecha aprox: {row['datetime_str']}")
print(f"  |r| = {r_dart_mag:.4f} AU")
print(f"  |v| = {v_dart_mag * UV_ms:.4f} m/s  ({v_dart_mag:.6f} AU/UT)")
print(f"  ε   = {eps_before:.8f} AU²/UT²")
print(f"  a   = {a_before:.6f} AU")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# ── ERE jump ────────────────────────────────────────────────────────────────
axes[0, 0].plot(dv_range_ms,
                (eps_after_rad - eps_before) / abs(eps_before) * 100,
                'b-', lw=2, label='impulso radial')
axes[0, 0].plot(dv_range_ms,
                (eps_after_tan - eps_before) / abs(eps_before) * 100,
                'r--', lw=2, label='impulso tangencial')
axes[0, 0].set_xlabel(r'$|\Delta V|$ [m/s]', fontsize=11)
axes[0, 0].set_ylabel(r'$\Delta\varepsilon / |\varepsilon_0|$ [%]', fontsize=11)
axes[0, 0].set_title('Salto en ERE vs magnitud del ΔV', fontsize=11)
axes[0, 0].legend(fontsize=9)
axes[0, 0].grid(True, alpha=0.3)

# ── New semi-major axis ─────────────────────────────────────────────────────
axes[0, 1].plot(dv_range_ms, a_after_rad,  'b-',  lw=2, label='impulso radial')
axes[0, 1].plot(dv_range_ms, a_after_tan,  'r--', lw=2, label='impulso tangencial')
axes[0, 1].axhline(a_before, color='k', ls='--', lw=1.2, label=f'a₀ = {a_before:.4f} AU')
axes[0, 1].set_xlabel(r'$|\Delta V|$ [m/s]', fontsize=11)
axes[0, 1].set_ylabel('Semieje mayor $a$ [AU]', fontsize=11)
axes[0, 1].set_title('Cambio de semieje mayor (vis-viva)', fontsize=11)
axes[0, 1].legend(fontsize=9)
axes[0, 1].grid(True, alpha=0.3)

# ── Δh/h₀ for radial ────────────────────────────────────────────────────────
axes[1, 0].plot(dv_range_ms, dh_mag_rad * 100, 'steelblue', lw=2, label='impulso radial')
axes[1, 0].set_xlabel(r'$|\Delta V|$ [m/s]', fontsize=11)
axes[1, 0].set_ylabel(r'$\Delta|\mathbf{h}| / h_0$ [%]', fontsize=11)
axes[1, 0].set_title('Cambio en MARE vs ΔV radial', fontsize=11)
axes[1, 0].legend(fontsize=9)
axes[1, 0].grid(True, alpha=0.3)

# Note: radial impulse → r×ΔV is NOT zero (angle with r)
axes[1, 0].annotate(
    r'Impulso radial: $\mathbf{r}\times\Delta\mathbf{V}=0$ si $\Delta\mathbf{V}\parallel\hat{r}$\n'
    r'→ MARE no cambia para ΔV puramente radial',
    xy=(0.5, 0.6), xycoords='axes fraction', fontsize=9,
    ha='center', va='center',
    bbox=dict(boxstyle='round,pad=0.3', fc='lightyellow', ec='goldenrod', alpha=0.9)
)

# ── Δe_LRL for radial ───────────────────────────────────────────────────────
axes[1, 1].plot(dv_range_ms, de_lrl_rad * 100, 'seagreen', lw=2, label='impulso radial')
axes[1, 1].set_xlabel(r'$|\Delta V|$ [m/s]', fontsize=11)
axes[1, 1].set_ylabel(r'$\Delta|\mathbf{e}_{LRL}| / e_0$ [%]', fontsize=11)
axes[1, 1].set_title('Cambio en vector LRL (excentricidad) vs ΔV radial', fontsize=11)
axes[1, 1].legend(fontsize=9)
axes[1, 1].grid(True, alpha=0.3)

plt.suptitle(
    f'Salto instantáneo en invariantes al aplicar ΔV (τ = 180 d antes del encuentro)',
    fontsize=12
)
plt.tight_layout()
plt.show()
```

## 8. Comparación 2-Cuerpos vs N-Cuerpos: Efecto de la Perturbación Terrestre

Para evidenciar la diferencia entre la conservación **ideal** (dos cuerpos Sol-Apophis) y la
conservación **real** (tres cuerpos Sol + Tierra + Apophis), integramos ambos sistemas
numéricamente con `pymcel` durante los 90 días alrededor del encuentro 2029:

| Sistema | Conservación esperada | Función usada |
|---------|----------------------|---------------|
| 2 cuerpos (Sol + Apophis) | $\\sim 10^{-12}$ (solo error numérico) | `pc.doscuerpos_solucion` |
| 3 cuerpos (Sol + Tierra + Apophis) | Se rompe cerca del encuentro | `pc.ncuerpos_solucion` |

Las condiciones iniciales provienen de las efemérides de Horizons evaluadas 45 días antes
del encuentro ($\\approx$ 2029-02-27).

```python
# ── 45 days before encounter: 2029-02-27 ────────────────────────────────────
idx_start = np.argmin(np.abs(dates - pd.Timestamp('2029-02-27')))
row0 = df.iloc[idx_start]

r0_aph = np.array([row0['x'],      row0['y'],      row0['z']])
v0_aph = np.array([row0['vx_can'], row0['vy_can'], row0['vz_can']])

# Integration time: 90 days centered on encounter
t_span = 90.0 / UT_days   # 90 days in UT
ts_nbody = np.linspace(0, t_span, 1000)

# ── 2-BODY: Sun + Apophis ────────────────────────────────────────────────────
rs_2b, vs_2b = pc.doscuerpos_solucion(mu_sol, r0_aph, v0_aph, ts_nbody)

h_2b  = np.cross(rs_2b, vs_2b)
h_2b_mag = np.linalg.norm(h_2b, axis=1)
r_2b_mag = np.linalg.norm(rs_2b, axis=1)
v_2b_mag = np.linalg.norm(vs_2b, axis=1)
eps_2b   = v_2b_mag**2 / 2 - mu_sol / r_2b_mag
r_2b_hat = rs_2b / r_2b_mag[:, None]
e_2b_vec = np.cross(vs_2b, h_2b) / mu_sol - r_2b_hat
e_2b_mag = np.linalg.norm(e_2b_vec, axis=1)

print("2-BODY (Sol + Apophis) — errores de conservación:")
print(f"  |Δh/h₀|_max = {np.abs((h_2b_mag - h_2b_mag[0])/h_2b_mag[0]).max():.2e}")
print(f"  |Δε/ε₀|_max = {np.abs((eps_2b - eps_2b[0])/abs(eps_2b[0])).max():.2e}")
print(f"  |Δe/e₀|_max = {np.abs((e_2b_mag - e_2b_mag[0])/e_2b_mag[0]).max():.2e}")
```

```python
# ── Fetch Earth state at the same epoch from Horizons ───────────────────────
epoch_start_str = str(dates.iloc[idx_start].date())
epoch_end_str   = str((dates.iloc[idx_start] + pd.Timedelta(days=91)).date())

df_earth = pc.consulta_horizons(
    id='399',
    location='@0',
    epochs={'start': epoch_start_str, 'stop': epoch_end_str, 'step': '1d'},
    datos='vectors'
)

r0_earth = np.array([df_earth.iloc[0]['x'],      df_earth.iloc[0]['y'],      df_earth.iloc[0]['z']])
v0_earth = np.array([df_earth.iloc[0]['vx'] * UT_days,
                     df_earth.iloc[0]['vy'] * UT_days,
                     df_earth.iloc[0]['vz'] * UT_days])

# Mass ratio Tierra/Sol (canonical: m_Sol = 1)
from pymcel.constantes import GM_earth
m_earth_can = GM_earth / GM_sun

sistema_3b = [
    dict(m=1.0,            r=[0., 0., 0.], v=[0., 0., 0.]),   # Sol (centro)
    dict(m=m_earth_can,    r=list(r0_earth), v=list(v0_earth)), # Tierra
    dict(m=0.0,            r=list(r0_aph),   v=list(v0_aph)),  # Apophis (test particle)
]

# Integrate 3-body
rs_nb, vs_nb, rps_nb, vps_nb, const_nb = pc.ncuerpos_solucion(sistema_3b, ts_nbody)

# Apophis state in inertial frame (particle index 2)
rs_aph_nb = rs_nb[2]   # (N,3)
vs_aph_nb = vs_nb[2]   # (N,3)

h_nb  = np.cross(rs_aph_nb, vs_aph_nb)
h_nb_mag = np.linalg.norm(h_nb, axis=1)
r_nb_mag = np.linalg.norm(rs_aph_nb, axis=1)
v_nb_mag = np.linalg.norm(vs_aph_nb, axis=1)
eps_nb   = v_nb_mag**2 / 2 - mu_sol / r_nb_mag
r_nb_hat = rs_aph_nb / r_nb_mag[:, None]
e_nb_vec = np.cross(vs_aph_nb, h_nb) / mu_sol - r_nb_hat
e_nb_mag = np.linalg.norm(e_nb_vec, axis=1)

print("3-BODY (Sol + Tierra + Apophis) — errores de conservación:")
print(f"  |Δh/h₀|_max = {np.abs((h_nb_mag - h_nb_mag[0])/h_nb_mag[0]).max():.2e}")
print(f"  |Δε/ε₀|_max = {np.abs((eps_nb - eps_nb[0])/abs(eps_nb[0])).max():.2e}")
print(f"  |Δe/e₀|_max = {np.abs((e_nb_mag - e_nb_mag[0])/e_nb_mag[0]).max():.2e}")
```

## 9. Conclusiones

### 9.1 Conservación kepleriana verificada

Para las efemérides de Apophis consultadas a través de Horizons (que incorporan la fuerza
gravitacional del Sol como perturbación dominante), los tres invariantes keplerianos se
conservan al nivel de:

$$\\left|\\frac{\\Delta h}{h_0}\\right|_{\\max} < 10^{-6}, \\quad
\\left|\\frac{\\Delta\\varepsilon}{\\varepsilon_0}\\right|_{\\max} < 10^{-6}, \\quad
\\left|\\frac{\\Delta e}{e_0}\\right|_{\\max} < 10^{-6}$$

Las pequeñas desviaciones visibles reflejan las perturbaciones de Júpiter, Tierra y otros
cuerpos incluidos en la integración de Horizons, **no errores numéricos**.

---

### 9.2 Ruptura de invariantes por el impacto DART

Un impulso $\\Delta\\mathbf{V}$ en la dirección radial ($\\hat{\\mathbf{r}}$):

- **ERE:** cambia linealmente con $|\\Delta V|$ (término $\\mathbf{v}\\cdot\\Delta\\mathbf{V}$) + término
  cuadrático. Para $|\\Delta V| \\sim 100$ m/s, el cambio relativo en $\\varepsilon$ es $\\sim 1\\%$,
  lo que modifica el período orbital en $\\sim 1.5\\%$ (varias horas).

- **MARE:** el impulso radial satisface $\\mathbf{r} \\times \\hat{\\mathbf{r}} = \\mathbf{0}$,
  por lo que **no cambia $|\\mathbf{h}|$** en primer orden. Esto implica que la inclinación
  orbital y el nodo ascendente se mantienen.

- **LRL:** cambia su módulo (excentricidad) y dirección (periapsis). Para un DART inverso
  efectivo, la nueva dirección del LRL debe apuntar de modo que el periapsis quede **fuera**
  del tubo de riesgo de impacto con la Tierra.

---

### 9.3 Perturbación terrestre y conservación del LRL

La comparación 2-cuerpos vs 3-cuerpos muestra que la perturbación de la Tierra **rompe la
conservación del LRL** de forma notoria en la ventana $\\pm 30$ días del encuentro 2029:

$$\\left|\\frac{\\Delta e_{\\text{LRL}}}{e_0}\\right|_{\\text{3-cuerpos}} \\gg
\\left|\\frac{\\Delta e_{\\text{LRL}}}{e_0}\\right|_{\\text{2-cuerpos}}$$

Esto confirma que el LRL es el **indicador más sensible** de perturbaciones no keplerianas,
lo que lo hace una herramienta diagnóstica valiosa para detectar el efecto del impacto DART
en el análisis post-maniobra.

---

### 9.4 Implicaciones para la misión DART inverso

| Observación | Implicación para el diseño de la misión |
|-------------|----------------------------------------|
| $\\varepsilon$ cambia con $\\Delta V$ | Diseñar $|\\Delta V|$ para el $\\Delta a$ requerido |
| MARE insensible a impulso radial | Impulso tangencial más eficiente para cambiar $h$ |
| LRL rota con el impacto | Controlar dirección de $\\Delta\\mathbf{V}$ para apuntar $\\mathbf{e}'$ lejos de la Tierra |
| Perturbación terrestre modifica LRL | La maniobra debe ejecutarse con $\\tau \\geq 30$ días de antelación |
